```python
#  IQR OUTLIER DETECTION

def detect_outliers_iqr(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound


for column in [
       "units_sold",
    "price",
    "revenue",
    "marketing_spend"
]:

    if column in df.columns:

        outliers, lower, upper = detect_outliers_iqr(
            df,
            column
        )

        print(f"OUTLIERS: {column}")
        print("Lower bound:", lower)
        print("Upper bound:", upper)
        print("Number of outliers:", len(outliers))
```

```python
#feature engineering

df = df.sort_values(["product_id", "store_id", "date"]).reset_index(drop=True)

TARGET = "units_sold"
GROUP_COLS = ["product_id", "store_id"] 
 
df["year"] = df["date"].dt.year
print(df["year"])

df["month_number"] = df["date"].dt.month

df["day"] = df["date"].dt.day

df["day_number"] = df["date"].dt.dayofweek

df["week_number"] = (
    df["date"]
    .dt.isocalendar()
    .week
    .astype(int)
)
```

```python
# STEP 1: Time-series feature -> is_weekend
#Formula (from doc): 1 for Saturday/Sunday, otherwise 0

df["is_weekend"] = df["date"].dt.dayofweek.isin([5, 6]).astype(int)
print(df[["date", "is_weekend"]])
```

```python
#STEP 2: Time-series feature -> sales_lag_7
# Formula (from doc): Sales from seven days earlier (same product+store)

df["sales_lag_7"] = df.groupby(GROUP_COLS)[TARGET].shift(7)
df["sales_avg_30d"] = (
    df.groupby(GROUP_COLS)[TARGET]
      .transform(lambda s: s.shift(1).rolling(window=30, min_periods=1).mean())
)
```

```python
# STEP 4: Derived feature -> discount_ratio
# Formula (from doc): discount_ratio = discount amount / original price
# Dataset stores Discount_Percentage, so first compute the discount amount
# ==================================================================
df["discount_amount"] = df["price"] * (df["discount_percentage"] / 100)
df["discount_ratio"] = np.where(df["price"] > 0, df["discount_amount"] / df["price"], 0)
```

```python
df["sales_lag_7"] = df["sales_lag_7"].fillna(0)
df["sales_avg_30d"] = df["sales_avg_30d"].fillna(0)
```

```python
outlier_cols = [
    "price",
    "discount_percentage",
    "marketing_spend",
    "competitor_price",
    "sales_avg_30d"
]
for col in outlier_cols:
    Q1, Q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower, upper) 
```

```python
# STEP 2C: Create practical lag features for sparse data

df_model = df.copy()

df_model["date"] = pd.to_datetime(df_model["date"])

# Sort correctly
df_model = df_model.sort_values(
    ["product_id", "store_id", "date"]
).reset_index(drop=True)

print("Total rows:", len(df_model))
print("Date range:",
      df_model["date"].min(),
      "to",
      df_model["date"].max())


# Create record-based lag features
group_cols = ["product_id", "store_id"]

df_model["sales_lag_1"] = (
    df_model.groupby(group_cols)["units_sold"]
    .shift(1)
)

df_model["sales_lag_2"] = (
    df_model.groupby(group_cols)["units_sold"]
    .shift(2)
)

df_model["sales_lag_3"] = (
    df_model.groupby(group_cols)["units_sold"]
    .shift(3)
)

df_model["sales_lag_7_records"] = (
    df_model.groupby(group_cols)["units_sold"]
    .shift(7)
)

# Previous-record rolling averages
df_model["sales_avg_3"] = (
    df_model.groupby(group_cols)["units_sold"]
    .transform(
        lambda s: s.shift(1)
        .rolling(window=3, min_periods=1)
        .mean()
    )
)

df_model["sales_avg_7"] = (
    df_model.groupby(group_cols)["units_sold"]
    .transform(
        lambda s: s.shift(1)
        .rolling(window=7, min_periods=1)
        .mean()
    )
)
#----------
df_model["discount_amount"] = (
    df_model["price"] *
    df_model["discount_percentage"] / 100
)

df_model["discount_ratio"] = (
    df_model["discount_percentage"] / 100
)


df_model = df_model.dropna(
    subset=[
        "sales_lag_1",
        "sales_lag_2",
        "sales_lag_3",
        "sales_lag_7_records"
    ]
).reset_index(drop=True)



print(
    df_model[
        [
            "date",
            "product_id",
            "store_id",
            "units_sold",
            "sales_lag_1",
            "sales_lag_2",
            "sales_lag_3",
            "sales_lag_7_records",
            "sales_avg_3",
            "sales_avg_7"
        ]
    ].head(20)
)

# ============================================================
# Create One-Hot Encoded Categorical Features
# ============================================================

categorical_columns = [
    "category",
    "store_location",
    "season",
    "weather",
    "sales_channel",
    "customer_segment",
    "day_of_week"
]

df_model = pd.get_dummies(
    df_model,
    columns=categorical_columns,
    prefix=categorical_columns,
    dtype=int
)

print("One-hot encoding completed")
print(df_model.columns.tolist())

# Check whether any required features are still missing

missing_features = [
    col for col in FEATURES
    if col not in df_model.columns
]

print("\nMissing features:")
print(missing_features)


# ============================================================
# STEP 2.5: Convert Quarter to Numeric
# ============================================================

print("Quarter before conversion:")
print(df_model["quarter"].unique())

df_model["quarter"] = (
    df_model["quarter"]
    .astype(str)
    .str.replace("Q", "", regex=False)
    .astype(int)
)

print("\nQuarter after conversion:")
print(df_model["quarter"].unique())
```

```python
# STEP 3: Prepare final features for model training

TARGET = "units_sold"

FEATURES = [
    # Historical sales features
    "sales_lag_1",
    "sales_lag_2",
    "sales_lag_3",
    "sales_lag_7_records",
    "sales_avg_3",
    "sales_avg_7",

    # Business features
    "price",
    "discount_percentage",
    "promotion_flag",
    "stock_availability",
    "holiday_flag",
    "local_event_flag",
    "competitor_price",
    "economic_indicator",
    "marketing_spend",
    "discount_amount",
    "discount_ratio",

    # Time features
    "quarter",
    "is_weekend",
    "year",
    "month_number",
    "day",
    "day_number",
    "week_number",

    # Category
    "category_Appliances",
    "category_Electronics",
    "category_Furniture",
    "category_Home",
    "category_Sports",
    "category_Stationery",

    # Store location
    "store_location_Chennai",
    "store_location_Delhi",
    "store_location_Mumbai",

    # Season
    "season_Spring",
    "season_Summer",
    "season_Winter",

    # Weather
    "weather_Cloudy",
    "weather_Rainy",
    "weather_Stormy",
    "weather_Sunny",

    # Sales channel
    "sales_channel_Online",
    "sales_channel_Wholesale",

    # Customer segment
    "customer_segment_Retail",
    "customer_segment_Wholesale",

    # Day of week
    "day_of_week_Monday",
    "day_of_week_Saturday",
    "day_of_week_Sunday",
    "day_of_week_Thursday",
    "day_of_week_Tuesday",
    "day_of_week_Wednesday"
]

print("Target:", TARGET)
print("Number of features:", len(FEATURES))

# Check whether any requested features are missing
missing_features = [
    col for col in FEATURES
    if col not in df_model.columns
]

print("Missing features:", missing_features)
```

```python
# STEP 4: Chronological Train / Validation / Test Split

# Make sure data is sorted by date
df_model = df_model.sort_values("date").reset_index(drop=True)

# Define X and y
X = df_model[FEATURES].copy()
y = df_model[TARGET].copy()

# Check for non-numeric columns

non_numeric = X.select_dtypes(
    exclude=np.number
).columns

print("Non-numeric columns:")
print(non_numeric)


# Use date to create chronological splits
train_end = pd.Timestamp("2025-06-30")
validation_end = pd.Timestamp("2025-09-30")

train_mask = df_model["date"] <= train_end

validation_mask = (
    (df_model["date"] > train_end) &
    (df_model["date"] <= validation_end)
)

test_mask = df_model["date"] > validation_end

X_train = X.loc[train_mask].copy()
y_train = y.loc[train_mask].copy()

X_val = X.loc[validation_mask].copy()
y_val = y.loc[validation_mask].copy()

X_test = X.loc[test_mask].copy()
y_test = y.loc[test_mask].copy()

print("TRAIN")
print("Rows:", len(X_train))
print("Date:", df_model.loc[train_mask, "date"].min(),
      "to",
      df_model.loc[train_mask, "date"].max())

print("\nVALIDATION")
print("Rows:", len(X_val))
print("Date:", df_model.loc[validation_mask, "date"].min(),
      "to",
      df_model.loc[validation_mask, "date"].max())

print("\nTEST")
print("Rows:", len(X_test))
print("Date:", df_model.loc[test_mask, "date"].min(),
      "to",
      df_model.loc[test_mask, "date"].max())
```

```python
# STEP 6: Baseline model
# Predict today's sales using the previous available sales

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Baseline prediction = previous available sales
baseline_val_pred = X_val["sales_lag_1"].copy()
baseline_test_pred = X_test["sales_lag_1"].copy()

# Remove rows where lag is missing
val_valid = baseline_val_pred.notna()
test_valid = baseline_test_pred.notna()

# Validation metrics
baseline_val_mae = mean_absolute_error(
    y_val[val_valid],
    baseline_val_pred[val_valid]
)

baseline_val_rmse = np.sqrt(
    mean_squared_error(
        y_val[val_valid],
        baseline_val_pred[val_valid]
    )
)

# Test metrics
baseline_test_mae = mean_absolute_error(
    y_test[test_valid],
    baseline_test_pred[test_valid]
)

baseline_test_rmse = np.sqrt(
    mean_squared_error(
        y_test[test_valid],
        baseline_test_pred[test_valid]
    )
)

print("BASELINE RESULTS")
print("----------------")

print("Validation MAE :", round(baseline_val_mae, 2))
print("Validation RMSE:", round(baseline_val_rmse, 2))

print("Test MAE       :", round(baseline_test_mae, 2))
print("Test RMSE      :", round(baseline_test_rmse, 2))

print("\nValidation rows used:", val_valid.sum())
print("Test rows used:", test_valid.sum())
```

```python
models = {

     "XGBoost": XGBRegressor(
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        objective="reg:squarederror"
    )
}

results = []

for name, model in models.items():

    print("\nTraining:", name)

    # Train
    model.fit(X_train, y_train)

    # Validation prediction
    val_pred = model.predict(X_val)

    # Validation metrics
    val_mae = mean_absolute_error(y_val, val_pred)

    val_rmse = np.sqrt(
        mean_squared_error(y_val, val_pred)
    )

    # Store results
    results.append({
        "Model": name,
        "Validation MAE": val_mae,
        "Validation RMSE": val_rmse
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "Validation MAE"
).reset_index(drop=True)
print("\n==============================")
print("MODEL COMPARISON")
print("==============================")

print(results_df)

# ============================================================
# STEP 9: PREPARE TIME-SERIES DATA


print("\n========================================")
print("STEP 9: PREPARE TIME-SERIES DATA")
print("========================================")

# Make sure date is datetime
df_model["date"] = pd.to_datetime(df_model["date"])

# Aggregate total units sold for each date
daily_sales = (
    df_model
    .groupby("date")["units_sold"]
    .sum()
    .reset_index()
    .sort_values("date")
    .reset_index(drop=True)
)

print("\nDaily sales data:")
print(daily_sales.head(10))

print("\nNumber of daily records:", len(daily_sales))

print(
    "\nDate range:",
    daily_sales["date"].min(),
    "to",
    daily_sales["date"].max()
)
# ============================================================
# STEP 10: TIME-SERIES TRAIN / VALIDATION SPLIT
# ============================================================

print("\n========================================")
print("STEP 10: TIME-SERIES SPLIT")
print("========================================")

# # Use 80% for training
# split_index = int(len(daily_sales) * 0.8)

# ts_train = daily_sales.iloc[:split_index].copy()

# ts_val = daily_sales.iloc[split_index:].copy()
# Same dates used by ML models
train_end = pd.Timestamp("2025-06-30")
validation_end = pd.Timestamp("2025-09-30")

# Time-series training data
ts_train = daily_sales[
    daily_sales["date"] <= train_end
].copy()

# Time-series validation data
ts_val = daily_sales[
    (daily_sales["date"] > train_end) &
    (daily_sales["date"] <= validation_end)
].copy()

# Time-series test data
ts_test = daily_sales[
    daily_sales["date"] > validation_end
].copy()


print("\nTIME-SERIES TRAIN")
print("Rows:", len(ts_train))
print(
    "Date:",
    ts_train["date"].min(),
    "to",
    ts_train["date"].max()
)

print("\nTIME-SERIES VALIDATION")
print("Rows:", len(ts_val))
print(
    "Date:",
    ts_val["date"].min(),
    "to",
    ts_val["date"].max()
)

print("\nTIME-SERIES TEST")
print("Rows:", len(ts_test))
print(
    "Date:",
    ts_test["date"].min(),
    "to",
    ts_test["date"].max()
)

# STEP 11: ARIMA MODEL
# ============================================================

print("\n========================================")
print("STEP 11: TRAINING ARIMA")
print("========================================")

from statsmodels.tsa.arima.model import ARIMA

# Training data
arima_train = ts_train["units_sold"]

# Create ARIMA model
arima_model = ARIMA(
    arima_train,
    order=(1, 1, 1)
)

# Train model
arima_fit = arima_model.fit()

# Forecast validation period
arima_pred = arima_fit.forecast(
    steps=len(ts_val)
)

# Calculate MAE
arima_mae = mean_absolute_error(
    ts_val["units_sold"],
    arima_pred
)

# Calculate RMSE
arima_rmse = np.sqrt(
    mean_squared_error(
        ts_val["units_sold"],
        arima_pred
    )
)

print("\nARIMA Results")
print("------------------------------")
print("MAE :", round(arima_mae, 4))
print("RMSE:", round(arima_rmse, 4))
# ============================================================
# STEP 12: PROPHET MODEL
# ============================================================

print("\n========================================")
print("STEP 12: TRAINING PROPHET")
print("========================================")

from prophet import Prophet

# Prophet requires columns named ds and y
prophet_train = ts_train.rename(
    columns={
        "date": "ds",
        "units_sold": "y"
    }
)

prophet_val = ts_val.rename(
    columns={
        "date": "ds",
        "units_sold": "y"
    }
)

# Create Prophet model
prophet_model = Prophet()

# Train
prophet_model.fit(prophet_train)

# Create future dates
future = prophet_model.make_future_dataframe(
    periods=len(prophet_val),
    freq="D"
)

# Forecast
forecast = prophet_model.predict(future)

# Get validation predictions
prophet_pred = (
    forecast["yhat"]
    .iloc[-len(prophet_val):]
    .values
)

# Calculate MAE
prophet_mae = mean_absolute_error(
    prophet_val["y"],
    prophet_pred
)

# Calculate RMSE
prophet_rmse = np.sqrt(
    mean_squared_error(
        prophet_val["y"],
        prophet_pred
    )
)

print("\nProphet Results")
print("------------------------------")
print("MAE :", round(prophet_mae, 4))
print("RMSE:", round(prophet_rmse, 4))

# STEP 13: LSTM MODEL
# ============================================================

print("\n========================================")
print("STEP 13: TRAINING LSTM")
print("========================================")

# ============================================================
# STEP 13: LSTM
# ============================================================

print("\n========================================")
print("STEP 13: TRAINING LSTM")
print("========================================")

from sklearn.preprocessing import MinMaxScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Input

# ------------------------------------------------------------
# Prepare training and validation values
# ------------------------------------------------------------

train_values = ts_train["units_sold"].values.reshape(-1, 1)

val_values = ts_val["units_sold"].values.reshape(-1, 1)

# ------------------------------------------------------------
# Scale using TRAINING data only
# ------------------------------------------------------------

scaler = MinMaxScaler()

train_scaled = scaler.fit_transform(
    train_values
)

val_scaled = scaler.transform(
    val_values
)

# ------------------------------------------------------------
# Create sequences
# ------------------------------------------------------------

LOOKBACK = 7


def create_sequences(data, lookback):

    X = []
    y = []

    for i in range(
        lookback,
        len(data)
    ):

        X.append(
            data[i - lookback:i, 0]
        )

        y.append(
            data[i, 0]
        )

    return np.array(X), np.array(y)


# Training sequences
X_lstm_train, y_lstm_train = create_sequences(
    train_scaled,
    LOOKBACK
)

# ------------------------------------------------------------
# Validation sequences
# ------------------------------------------------------------

# Include last LOOKBACK training observations
# so validation can use previous history

combined_scaled = np.concatenate(
    [
        train_scaled[-LOOKBACK:],
        val_scaled
    ]
)

X_lstm_val, y_lstm_val = create_sequences(
    combined_scaled,
    LOOKBACK
)

# ------------------------------------------------------------
# Reshape
# ------------------------------------------------------------

X_lstm_train = X_lstm_train.reshape(
    X_lstm_train.shape[0],
    X_lstm_train.shape[1],
    1
)

X_lstm_val = X_lstm_val.reshape(
    X_lstm_val.shape[0],
    X_lstm_val.shape[1],
    1
)

print(
    "LSTM training samples:",
    len(X_lstm_train)
)

print(
    "LSTM validation samples:",
    len(X_lstm_val)
)

# ------------------------------------------------------------
# Build model
# ------------------------------------------------------------

lstm_model = Sequential([
    Input(shape=(LOOKBACK, 1)),
    LSTM(50),
    Dense(1)
])

# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

lstm_model.compile(
    optimizer="adam",
    loss="mse"
)

# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

lstm_model.fit(
    X_lstm_train,
    y_lstm_train,
    epochs=20,
    batch_size=32,
    verbose=1
)

# ------------------------------------------------------------
# Predict validation
# ------------------------------------------------------------

lstm_pred_scaled = lstm_model.predict(
    X_lstm_val,
    verbose=0
)

# Convert back to original scale

lstm_pred = scaler.inverse_transform(
    lstm_pred_scaled
).flatten()

y_lstm_actual = scaler.inverse_transform(
    y_lstm_val.reshape(-1, 1)
).flatten()

# ------------------------------------------------------------
# Evaluation
# ------------------------------------------------------------

lstm_mae = mean_absolute_error(
    y_lstm_actual,
    lstm_pred
)

lstm_rmse = np.sqrt(
    mean_squared_error(
        y_lstm_actual,
        lstm_pred
    )
)

print("\nLSTM Results")
print("------------------------------")
print("MAE :", round(lstm_mae, 4))
print("RMSE:", round(lstm_rmse, 4))
```
